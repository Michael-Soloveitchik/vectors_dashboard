# vectors_dashboard

Linked explorer for embedding vectors: projection · pair / centroid / confusion matrix / class tree · ROC / DET · per-class score distribution · click-to-open inspector (WavLens audio when wavs exist).
Every action is a transaction: the top bar shows `● in sync` once all panels agree.

1. paths + drop cached dashboard modules (code edits are picked up without a kernel restart)
2. build the `Gui` — `USE_REAL = False` synthetic demo, `True` real embeddings + audio
3. run — shown **here in the notebook** and at the printed URL(s) for a browser tab. Re-running stops the previous server on `PORT` and rebinds; a port held by another process is left alone (the next free port is used)

Mouse: click = inspect · Ctrl+click = add / remove from selection · drag = select · Shift+drag = add · double-click = clear selection · right-click or ⌂ = reset zoom · `x` = clear selection · `Esc` = close inspector.

```python
Gui.demo()                                   # synthetic
Gui.real("idrnd_redimnet_b3")                # real parquet(s) + wavs
```

````python
PORT = 8070

import sys
from pathlib import Path

# Project root = the folder holding code/frontend: the working directory, a
# parent of it, or (VS Code) the folder of this notebook file.
_here = [Path.cwd(), *Path.cwd().parents]
if '__vsc_ipynb_file__' in globals():
    _here.insert(0, Path(globals()['__vsc_ipynb_file__']).parent)
ROOT = next((p for p in _here if (p / 'code' / 'frontend').is_dir()), None)
if ROOT is None:
    raise RuntimeError('run this notebook from the vectors_dashboard folder')
for p in (str(ROOT), str(ROOT / 'code')):
    if p not in sys.path:
        sys.path.insert(0, p)

# Drop cached dashboard modules so re-running picks up code edits.
for _m in list(sys.modules):
    if _m.split('.')[0] in ('frontend', 'plot', 'backend', 'data'):
        del sys.modules[_m]
````
